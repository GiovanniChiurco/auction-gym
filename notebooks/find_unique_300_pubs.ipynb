```python
import pandas as pd
import json
import os
```

## Carico dati da tutti gli esperimenti con Bid Shading

```python
res_dir_list = [
    'FP_DM_Oracle_sigmoids_cucb_maxucbclick_alpha_0',
    'FP_DM_Oracle_sigmoids_cucb_maxucbclick_alpha_1',
    'FP_DM_Oracle_sigmoids_sw_cucb_alpha_0_ws_70',
    'FP_DM_Oracle_sigmoids_sw_cucb_alpha_1_ws_50_70_100'
]
```

```python
all_files = []
for res_dir in res_dir_list:
    for file in os.listdir(f"../results/{res_dir}"):
        if file.startswith("agent"):
            all_files.append(os.path.join(res_dir, file))
```

```python
all_files_df = pd.concat(
    [pd.read_csv(f"../results/{file}", usecols=['publisher']) for file in all_files])
```

```python
no_dup = all_files_df.drop_duplicates()
```

```python
no_dup = sorted(no_dup['publisher'].to_list())
```

## Leggo i 300 pub già salvati

```python
with open('../results/FP_Truthful_Oracle_sigmoids_cucb_est_click_impr_alphatune/300pubs.json', 'r') as f:
    pubs = json.load(f)
```

```python
pubs = sorted(pubs)
```

```python
print(no_dup == pubs)
```

## Mi assicuro che ho simulato una run per ognuno dei 300

```python
onerunperpub_dir_path = '../results/FP_DM_Oracle_sigmoids_1run1pub/'
onerunperpub_files = os.listdir(onerunperpub_dir_path)
onerunperpub_df = pd.concat(
    [pd.read_csv(f"{onerunperpub_dir_path}{file}", usecols=['publisher']) for file in onerunperpub_files]
)
```

```python
onerunperpub_df_no_dup = onerunperpub_df.drop_duplicates()
onerunperpub_pub_list = sorted(onerunperpub_df_no_dup['publisher'].to_list())
```

```python
common_pubs = [pub for pub in pubs if pub in onerunperpub_pub_list]
```

```python
print(len(common_pubs) == len(pubs))
```

## Salvo i publisher da simulare
