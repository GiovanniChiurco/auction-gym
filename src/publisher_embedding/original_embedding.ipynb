```python
import pickle
import numpy as np
```

```python
rng = np.random.default_rng(0)
```

```python
embedding_size = 5
embedding_var = 1.0
```

## Text Embedding 

```python
adv_embeddings = pickle.load(open("data/embeddings_to_pick/ad_embeddings_red_70.pkl", "rb"))
publisher_embeddings = pickle.load(open("data/embeddings_to_pick/sites_embeddings_red_70.pkl", "rb"))
```

```python
adv_names = list(adv_embeddings.keys())
publisher_names = list(publisher_embeddings.keys())
```

## Generate Gaussian Embeddings

```python
def generate_pub_gauss_embedding(publisher_name_list):
    pub_gaus_emb = {}
    for pub in publisher_name_list:
        pub_gaus_emb[pub] = np.concatenate((rng.normal(0, embedding_var, size=embedding_size), [1.0]))
    return pub_gaus_emb
```

```python
def generate_adv_gauss_embedding(adv_name_list):
    adv_gaus_emb = {}
    for adv in adv_name_list:
        curr_adv_emb = rng.normal(0.0, embedding_var, size=embedding_size)
        curr_adv_emb = np.hstack((curr_adv_emb, - 3.0 - 1.0 * rng.random(1)))
        adv_gaus_emb[adv] = curr_adv_emb
    return adv_gaus_emb
```

```python
pub_gaus_emb = generate_pub_gauss_embedding(publisher_names)
```

```python
adv_gaus_emb = generate_adv_gauss_embedding(adv_names)
```

```python
# Save the embeddings
with open(f"data/embeddings_to_pick/pub_gaus_emb_dim_{embedding_size}.pkl", "wb") as f:
    pickle.dump(pub_gaus_emb, f)
```

```python
with open(f"data/embeddings_to_pick/adv_gaus_emb_dim_{embedding_size}.pkl", "wb") as f:
    pickle.dump(adv_gaus_emb, f)
```
